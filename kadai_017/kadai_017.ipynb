```python
class Human:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age >= 20:
            print(f"{self.name}は大人です")
        else:
            print(f"{self.name}は大人ではありません")


humans = [
    Human("山田太郎", 25),
    Human("鈴木花子", 17),
    Human("佐藤次郎", 20),
]

for human in humans:
    human.check_adult()
```
